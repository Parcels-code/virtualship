# Your Virtual Expedition Across the North Sea

Now let’s get started on running your VirtualShip expedition. Follow the steps below to plan your expedition, and launch your simulation.

### Step 1. Register with the Copernicus Marine Data Store
***

You will need to register for **Copernicus Marine Service** account (see [here](https://data.marine.copernicus.eu/register)) as VirtualShip streams ocean data directly from there. Register and remember your username and password before you start, because you will be asked for your credentials the first time you run an expedition.

### Step 2. Open your GitHub Codespaces Environment and Upload your Cruise Plan
***

In your Codespaces environment, upload the cruise coordinate excel file. 

From now you will work within the Terminal. You can find the Terminal window in the bottom half of your screen ([see here for what it should look like](https://virtualship.readthedocs.io/en/latest/user-guide/tutorials/codespaces_guide.html#using-the-virtualship-simulation-workspace))

We suggest practicing good file management and creating a dedicated folder for your group's expedition data in the workspace. You can do so by running `mkdir {your-group-name}` in the Terminal (replacing `{your-group-name}` with your actual group name) or via the File Explore panel in the Codespaces interface.

Navigate to the directory where your group will be running your expedition. Then upload the cruise file by right clicking in the File Explorer panel (ideally in the directory you created for your group's expedition data) and selecting the "Upload" option. You can also simply drag and drop your file to the Codespaces environment. From here you will be able to access your excel file.

### Step 3. Expedition Initialisation
***

In the Terminal, navigate to the folder where you uploaded your cruise plan using `cd your-folder-name`. This is where you will run your expedition.

Now enter the following command in the Terminal. Make sure to change:
- `EXPEDITION_NAME` to something more meaningful for your group’s expedition (eg `Expedition_Group1`)
    - For all subsequent commands make sure to replace `EXPEDITION_NAME` with the name you have chosen
- Verify your excel file is called CruiseDataExport-Transect.xlsx, otherwise change it to what it is called
- replace 'YYYY-MM-DD HH:MM:SS' with your start date and time (eg '1997-05-01 08:00:00') (Use the choice you selected in the Excel file on the 19th)

```bash
virtualship init EXPEDITION_NAME --from-mfp CruiseDataExport-Transect.xlsx --start-date 'YYYY-MM-DD HH:MM:SS'
``` 

This will create a folder/directory called `EXPEDITION_NAME` (or what you have changed this to) with a single file: `expedition.yaml.` This file contains details on the ship and instrument configurations, as well as the expedition schedule based on the sampling site coordinates from your cruise file. 

### Step 4. Expedition Scheduling & Ship Configuration
***

The next step is to finalise the expedition schedule plan, including setting times and instrument selection choices for each waypoint. The easiest way to do so is to use the bespoke VirtualShip planning tool. Enter the following command in Terminal (replace EXPEDITION_NAME with the name you selected previously): 

```bash 
virtualship plan EXPEDITION_NAME
```

**Ship Speed**
- In the planning tool which appears, under Ship Config Editor > Ship Speed & Onboard Measurements, there is an option to change the ship speed. However, for this expedition, you should leave this as the default 10 knots value.

**Onboard Measurements**
- VirtualShip is capable of taking onboard temperature/salinity measurements and ADCP measurements. For your expedition, you may choose to turn on onboard temperature/salinity measurements if you would like but this is not required. 

**Instrument/Sensor Configuration**
- The most important instrument configuration setting to consider is the list of sensors for each instrument, which controls what type of measurements/variables the instrument records in the simulation and therefore what output data you will receive for each instrument. 

- For your expedition you will be working with **CTD**. Click on CTD and select a sensor you would like to use by toggling the respective switches on or off. Remember to only have **one sensor** switched on and this is the sensor you chose previously. (VirtualShip allows expeditions to run with multiple sensors and instruments, but for your research needs a single CTD sensor is most appropriate). 

- You do not need to make any other changes to instruments or ship configuration here. 

**Waypoint Datetimes**
- In the Waypoints Section, you will be able to see all your waypoints (sampling stations) as well as your port of departure and arrival. Based on the time you provided in 'YYYY-MM-DD HH:MM:SS' previously, the expedition will begin at that time and each waypoint time will already be pre-filled, based on estimated travel and station durations relative to your start date. 

- You will have a default of 45 minutes stationkeeping time at each station. This may be enough for your expedition, but it may not be, so expect to adjust some of your timings once you have chosen your CTD sensor. The ship needs to slow down to a stop, which takes 15 minutes (default in VirtualShip). The CTD is lowered to the configured maximum depth or the sea floor, whichever is shallowest. The time taken therefore depends on the depth in that location and the winch speed (default is 1 m/s in VirtualShip). Remember that the time is doubled as well because the CTD needs to be brought back up. (So for a depth of 100m, it will take 200 seconds for the CTD to go down and back up). The ship needs to speed back up again and set course to the next waypoint, which takes another 15 minutes (default in VirtualShip).

- It is important to ensure that the timings for each station are realistic. There must be enough time for the ship to travel to each site at the prescribed speed (10 knots), and enough time to actually carry out your deployments once you arrive. The expedition schedule will be automatically verified when you press Save Changes in the planning tool

- At each waypoint determine the time, and toggle on the switch for CTD to sample it at each waypoint. 

- Remember you have **9 days** for your expedition (from the moment of departure to arrival at port).  

**Save Changes**
- When you are happy with your ship configuration and schedule plan, press Save Changes at the bottom of the planning tool.

- On pressing Save Changes the tool will check the selections are valid (for example that the ship will be able to reach each waypoint in time). If they are, the changes will be saved to the expedition.yaml file, ready for the next steps. If your selections are invalid you should be provided with information on how to fix them.

### Step 5. Run the Expedition
***

The logistical planning is complete. Now you can now run your virtual expedition and hope for good weather! This stage will take all the measurements at each waypoint in your expedition schedule, using input data sourced from the [Copernicus Marine Data Store](https://data.marine.copernicus.eu/products). 

You will need to register for a Copernicus Marine Service account (you can do so [here](https://data.marine.copernicus.eu/register)), if you have not done so already. 

You can run your expedition simulation using the command (change `EXPEDITION_NAME` to the name you previously selected):
```bash
virtualship run EXPEDITION_NAME
```

If this is your first time running VirtualShip, you will be prompted to enter your own Copernicus Marine Data Store credentials (these will be saved automatically for future use). Enter your username (not email) and password. (*Note:* You will not see your password when entering it. This is normal.)

Small simulations (e.g. small space-time domains and fewer instrument deployments) will be relatively fast. For large, complex expeditions, it _could_ take up to an hour to simulate the measurements depending on your choices. Waiting for simulation is a great time to practice your level of patience. A skill much needed in oceanographic fieldwork!!

### Step 6. Results
***

Upon successfully completing the simulation, results from the expedition will be stored in the `EXPEDITION_NAME/results` directory, written in `.parquet` format. You can download the results file to your laptop by right clicking on the ctd.parquet file and selecting download. It is advised to download the file to a new folder on your laptop.

**Congratulations! You have finished your expedition across the North Sea!**

You have collected data from your research expedition. Now we will see how to analyse and understand your results.

**Interpreting your Results**
***

Here are some questions for you to consider to understand your results:
- Explain why the data is available at different depths throughout the expedition.
- Explain the pattern seen for your sensor as you move along the expedition. (Remember you start in the deep shelf and move closer to land (Texel) as the expedition progresses)

- For Oxygen Sensor: 
    - Describe the evolution of oxygen along the transect
    - Where is the lowest oxygen found and why do you think this is the case?
- For Nitrate Sensor:
    - Are there very high nitrate values nearest to land? If so, why do you think so?
- For pH Sensor:
    - Hypothesize why pH is lower in some spots.
- For Phytoplankton (/Chlorophyll) Sensor:
    - Where are the highest concentrations of chlorophyll and plankton?
