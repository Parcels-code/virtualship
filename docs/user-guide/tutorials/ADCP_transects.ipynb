## Data post-processing

Before we can continue, we need to do some post-processing to get it ready for plotting as a 2D distance × depth plot. Below are various helper functions which perform tasks such as calculating the ship's cumulative travel distance and calculating the various velocity components from the ADCP data.

```python
# utility functions


def haversine(lon1, lat1, lon2, lat2):
    """Great-circle distance (meters) between two points."""
    lon1, lat1, lon2, lat2 = map(np.radians, [lon1, lat1, lon2, lat2])
    dlon, dlat = lon2 - lon1, lat2 - lat1
    a = np.sin(dlat / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))
    return 6371000 * c


def distance_along_expedition(df):
    """Array of cumulative meters travelled along ADCP measurement locations."""
    d = np.zeros_like(df["x"], dtype=float)
    for ob in range(1, len(df["x"])):
        d[ob] = d[ob - 1] + haversine(
            df["x"][ob - 1], df["y"][ob - 1], df["x"][ob], df["y"][ob]
        )
    return d


def calc_velocities(df):
    """From U and V in df, calculate absolute, parallel, and perpendicular velocities, as well as compass flow direction, using a per-waypoint ship angle alpha."""
    traj_df = (
        df.select(["t", "x", "y"])
        .unique(subset=["t"])
        .sort("t")
        .with_columns(
            # next waypoint (x1, y1)
            pl.col("x").radians().alias("lon_rad"),
            pl.col("y").radians().alias("lat_rad"),
            pl.col("x").shift(-1).radians().alias("lon1_rad"),
            pl.col("y").shift(-1).radians().alias("lat1_rad"),
        )
        .with_columns(
            # forward differences to next waypoint
            (pl.col("lon1_rad") - pl.col("lon_rad")).alias("dlon"),
            (pl.col("lat1_rad") - pl.col("lat_rad")).alias("dlat"),
        )
        .with_columns(
            # alpha between (x, y) and (x1, y1) using arctan2
            pl.arctan2(pl.col("dlat"), pl.col("dlon") * pl.col("lat_rad").cos()).alias(
                "alpha"
            )
        )
        # fill final location alpha (where shift(-1) is Null) with backward fill
        .with_columns(pl.col("alpha").backward_fill())
        .select(["t", "alpha"])
    )

    # join per-timestep alpha back to main DataFrame and calculate velocity components
    return df.join(traj_df, on="t", how="left").with_columns(
        # Uabs = sqrt(U^2 + V^2)
        (pl.col("U").pow(2) + pl.col("V").pow(2)).sqrt().alias("Uabs"),
        # Uparallel = cos(alpha)*U + sin(alpha)*V
        (
            pl.col("alpha").cos() * pl.col("U") + pl.col("alpha").sin() * pl.col("V")
        ).alias("Uparallel"),
        # Uperp = -sin(alpha)*U + cos(alpha)*V
        (
            -pl.col("alpha").sin() * pl.col("U") + pl.col("alpha").cos() * pl.col("V")
        ).alias("Uperp"),
        # compass direction in degrees [0, 360)
        ((pl.arctan2(pl.col("U"), pl.col("V")).degrees() + 360) % 360).alias(
            "direction_deg"
        ),
    )


def get_velocity_2d_array(df, var_name, units):
    """Extract a single velocity variable as a 2D xarray DataArray (dims: t, z)."""
    # deduplicate (t, z) pairs in case of duplicate sampling
    df_clean = df.unique(subset=["t", "z"])

    # to 2D grid matrix (rows = t, columns = z)
    pivot_df = df_clean.pivot(on="z", index="t", values=var_name).sort("t")

    # map distance along track to each timestamp (t)
    dist_map = df_clean.select(["t", "distance"]).unique(subset=["t"]).sort("t")
    pivot_df = pivot_df.join(dist_map, on="t", how="left")

    # extract coordinates and matrix
    z_coords = sorted(
        [float(c) for c in pivot_df.columns if c not in ("t", "distance")]
    )
    dist_coords = pivot_df["distance"].to_numpy()
    t_coords = pivot_df["t"].to_numpy()

    data_matrix = pivot_df.select([pl.col(str(z)) for z in z_coords]).to_numpy()

    # create/return da
    return xr.DataArray(
        data=data_matrix,
        dims=["distance", "z"],
        coords={
            "distance": (
                "distance",
                dist_coords,
                {"long_name": "Distance along track"},
            ),
            "z": ("z", z_coords, {"units": "m", "long_name": "Depth"}),
            "t": (
                "distance",
                t_coords,
            ),  # Preserves time reference along distance dimension
        },
        name=var_name,
        attrs={"units": units, "long_name": f"{var_name} velocity component"},
    )
```


Now we will execute the various post-processing calculations, plus define some extra useful arrays to be used for the plotting...

```python
# distance from start as 1d array, and add as new column to adcp_df
distance_1d = distance_along_expedition(adcp_df)
adcp_df = adcp_df.with_columns(pl.Series("distance", distance_1d))

# calculate and add velocity components and direction to adcp_df
adcp_df = calc_velocities(adcp_df)

# convert velocity components to 2D xarray DataArrays (dims: distance, z)
Uabs = get_velocity_2d_array(adcp_df, "Uabs", units="m/s")
Uparallel = get_velocity_2d_array(adcp_df, "Uparallel", units="m/s")
Uperp = get_velocity_2d_array(adcp_df, "Uperp", units="m/s")
direction = get_velocity_2d_array(adcp_df, "direction_deg", units="degrees")

# coord arrays and landmask/seabed mask
distance = Uabs["distance"].values
depth = Uabs["z"].values
landmask = np.where(Uabs == 0, 1, np.nan)
```

## Plotting

We are now ready to plot our data. We will use distance travelled by the ship for the x-axis, and water column depth for the y-axis. The ADCP data will then be plotted according to the colour map for each diagnostic. The profiles across the trajectory are likely to be different depths because some parts of the ocean are of course shallower than others.

```python
PLOT_DICT = {
    "Uabs": {
        "data": Uabs,
        "cmap": cmo.speed,
        "label": "Absolute velocity",
        "norm": None,
    },
    "Uparallel": {
        "data": Uparallel,
        "cmap": cmo.balance,
        "label": "Along-track velocity (positive with flow)",
        "norm": mcolors.CenteredNorm(vcenter=0),
    },
    "Uperp": {
        "data": Uperp,
        "cmap": cmo.balance,
        "label": "Cross-track velocity (positive with flow to the left)",
        "norm": mcolors.CenteredNorm(vcenter=0),
    },
    "direction": {
        "data": direction,
        "cmap": cmo.phase,
        "label": "Flow direction (0=N, 90=E, 180=S, 270=W)",
        "norm": None,
    },
}

# fig
fig, axs = plt.subplots(
    len(PLOT_DICT), 1, figsize=(10, 10), dpi=96, sharex=True, sharey=True
)

for idx, ((key, var), ax) in enumerate(zip(PLOT_DICT.items(), axs, strict=False)):
    # adcp data
    mesh = ax.pcolormesh(
        distance / 1000,  # distance in km
        depth,
        var["data"].T.values,  # transpose to match (distance, z) dims
        cmap=var["cmap"],
        norm=var["norm"] if var["norm"] is not None else None,
    )

    # seabed
    ax.pcolormesh(
        distance / 1000,  # distance in km
        depth,
        landmask.T,
        cmap=mcolors.ListedColormap([mcolors.to_rgba("tan"), mcolors.to_rgba("white")]),
    )

    # title
    ax.set_title(var["label"])

    # colorbar
    if key == "direction":
        cbar = fig.colorbar(
            mesh,
            ax=ax,
            label="Degrees from North",
            ticks=[0, 90, 180, 270],
        )
    else:
        cbar = fig.colorbar(
            mesh,
            ax=ax,
            label=r"m s$^{-1}$",
        )

    # axis labels
    ax.set_ylabel("Depth (m)")
    if idx == len(axs) - 1:  # bottom panel only for single column of subplots
        ax.set_xlabel("Distance (km)")

# legend for sea bed
tan_patch = mpatches.Patch(color=mcolors.to_rgba("tan"), label="Land / sea bed")
axs[-1].legend(handles=[tan_patch], loc="lower right")


plt.tight_layout()
```

The resultant figure shows various components of the velocity field, derived from ADCP data.

1) Absolute velocity
2) Along-track velocity (where positive values indicate flow in the ship's travel direction between waypoints, i.e. the angle changes as the ship turns).
3) Cross-track velocity (where positive values indicate flow to the left of the ship's direction).
4) The direction of the flow, expressed as degrees from North.

You can use these plots as a starting point to consider how flow dynamics vary over the cross-section. You may find some diagnostics more useful than others, depending on your specific aims!
