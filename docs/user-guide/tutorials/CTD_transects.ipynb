# CTD Transect Plotting

This notebook demonstrates a simple plotting exercise for CTD data across a transect, using the output of a VirtualShip expedition. There are example plots embedded at the end, but these will ultimately be replaced by your own versions as you work through the notebook.

We can plot physical (temperature, salinity) or biogeochemical data (oxygen, chlorophyll, primary production, phytoplankton, nutrients, pH) as measured by the VirtualShip `CTD` instrument (dependent on which [sensors](../documentation/full_sensor_list.md) you have configured).

The plots we will produce are simple plots which follow the trajectory of the expedition as a function of distance travelled by the ship, and are intended to be a starting point for your analysis. 

<div class="alert alert-block alert-info"> 
<b>**Note**:</b> This notebook assumes that each waypoint in the expedition is further from the start than the last waypoint (i.e. waypoints arranged as a transect). The code will still work if not, but the resultant plots might not be very intuitive.
</div>

## Load data

```python
# load CTD data
ctd_df = parcels.read_particlefile(f"{data_dir}/ctd.parquet")
if np.unique(ctd_df["particle_id"]).size <= 1:
    raise ValueError("Number of waypoints must be > 1")
```

## Data post-processing

Before we can continue, we need to do some post-processing to get it ready for plotting as a 2D distance × depth plot. Below are various helper functions which perform tasks such as calculating the ship's cumulative travel distance, as well as some other utility methods. 

```python
# utility functions


def haversine(lon1, lat1, lon2, lat2):
    """Great-circle distance (meters) between two points."""
    lon1, lat1, lon2, lat2 = map(np.radians, [lon1, lat1, lon2, lat2])
    dlon, dlat = lon2 - lon1, lat2 - lat1
    a = np.sin(dlat / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))
    return 6371000 * c


def distance_along_expedition(df):
    """Array of cumulative meters travelled across CTD waypoints."""
    d = np.zeros_like(df["x"], dtype=float)
    for ob in range(1, len(df["x"])):
        d[ob] = d[ob - 1] + haversine(
            df["x"][ob - 1], df["y"][ob - 1], df["x"][ob], df["y"][ob]
        )
    return d


def ctd_build_2d_arrays(df, varname):
    """Build 2D (n_casts x uniform-depth) arrays from df. Ordered by their cumulative expedition distance."""

    # by cumulative distance
    cast_order = (
        df.group_by("particle_id").agg(pl.col("distance").first()).sort("distance")
    )

    profiles_z = []
    profiles_var = []
    cast_distances = []

    for row in cast_order.iter_rows(named=True):
        pid = row["particle_id"]
        cast = df.filter(pl.col("particle_id") == pid).sort(
            "z", descending=True
        )  # sort by depth (descending)
        profiles_z.append(cast["z"].to_numpy())
        profiles_var.append(cast[varname].to_numpy())
        cast_distances.append(row["distance"])

    # pad to uniform depth dimension with NaN
    max_len = max(len(p) for p in profiles_z)
    n_casts = len(profiles_z)

    z_2d = np.full((n_casts, max_len), np.nan)
    var_2d = np.full((n_casts, max_len), np.nan)

    for i, (pz, pv) in enumerate(zip(profiles_z, profiles_var)):
        z_2d[i, : len(pz)] = pz
        var_2d[i, : len(pv)] = pv

    return z_2d, np.array(cast_distances), var_2d


def build_masked_array(var_2d, profile_indices, n_profiles):
    arr = np.full((n_profiles, var_2d.shape[1]), np.nan)
    for i, idx in enumerate(profile_indices):
        if idx is not None:
            arr[i, :] = var_2d[idx, :]
    return arr


def get_profile_indices(distances):
    """Regular distance bins and profile indices for CTD transect plotting. Bin size is set to one order of magnitude lower than max distance."""
    dist_min, dist_max = float(np.min(distances)), float(np.max(distances))
    if dist_max > 1e6:
        dist_step = 1.0e5
    elif dist_max > 1e5:
        dist_step = 1.0e4
    elif dist_max > 1e4:
        dist_step = 1.0e3
    else:
        dist_step = 1.0e2  # fallback for very short transects

    distance_regular = np.arange(dist_min, dist_max + dist_step, dist_step)
    threshold = dist_step / 2
    profile_indices = [
        np.argmin(np.abs(distances - d))
        if np.min(np.abs(distances - d)) < threshold
        else None
        for d in distance_regular
    ]
    return profile_indices, distance_regular
```


Now we will execute the utility functions, plus define some extra useful arrays to be used for the plotting...

```python
# add distance along expedition
ctd_distance = distance_along_expedition(ctd_df)
ctd_df = ctd_df.with_columns(pl.Series("distance", ctd_distance))

# downcast only data
ctd_down = ctd_df.filter(pl.col("raising") == 0)

# build 2D arrays: (n_casts x uniform-depth)
z_2d, distances, var_2d = ctd_build_2d_arrays(ctd_down, plot_variable)

# 1D depth array from the deepest cast
deepest_idx = np.argmax(np.sum(~np.isnan(z_2d), axis=1))
z1d = z_2d[deepest_idx, :]
```

## Plotting

<div class="alert alert-block alert-info"> 
<b>**Note**:</b> The plots produced next are a starting point for your analysis. You are encouraged to make adjustments, for example axis limits and scaling if the defaults not best suited to your specific data.
</div>

We are now ready to plot our transect data. We will use distance travelled by the ship for the x-axis, and water column depth for the y-axis. The data for the chosen variable will then be plotted according to the colour map. The CTD casts are likely to be different depths because some parts of the ocean are of course shallower than others.

There are a few extra steps below which arrange the CTD casts into regular distance bins, so as to clearly demonstrate where along the transect we made CTD casts and indeed where there are gaps.


```python
# regularised transect
profile_indices, distance_regular = get_profile_indices(distances)
var_masked = build_masked_array(var_2d, profile_indices, len(distance_regular))

xticks_reg = np.linspace(
    float(distance_regular.min()),
    float(distance_regular.max()),
    len(distance_regular),
)

# plot regularised transect
fig, ax = plt.subplots(figsize=(10, 6), dpi=90)

ax.grid(True, which="both", color="lightgrey", linestyle="-", linewidth=0.7, alpha=0.5)

mesh = ax.pcolormesh(
    distance_regular / 1000,  # distance in km
    z1d,
    var_masked.T,
    cmap=VARIABLES[plot_variable]["cmap"],
)

ax.set_ylabel("Depth (m)")
ax.set_xlabel("Distance from start (km)")

plt.colorbar(mesh, ax=ax, label=VARIABLES[plot_variable]["label"])
plt.tight_layout()
```

In the plot above, we can see that there are gaps in the transects where no CTD casts have been made. After all, it's impossible to take measurements at every point across the transect! There will always be gaps making 10s of deployments across transects 1000s of kms long 🙃 This makes expedition/sampling site planning all the more important...

We can also also plot a 'filled' version without the distance bins, to give an alternative view of the evolution across the transect which is not dominated by gaps and white space. This time we will also add a 'sea bed' to the plot.

<div class="alert alert-block alert-info"> 
<b>**Note**:</b> It is important to remember that the gaps do actually exist in reality and this is a caveat which must be considered when interpreting the transect derived from CTD casts. Indeed, if you look at the x-axis of the plot below you will see that the deployments are not necessarily regularly spaced and some gaps are larger than others.
</div>

```python
# plot 'filled' transect (with sea bed visualised as well)
fig, ax = plt.subplots(figsize=(10, 6), dpi=96)

mesh = ax.pcolormesh(
    distances / 1000,  # distance in km
    z1d,
    var_2d.T,
    cmap=VARIABLES[plot_variable]["cmap"],
)

seabed = np.where(np.isnan(var_2d), 1, np.nan)  # sea bed
ax.pcolormesh(
    distances / 1000,  # distance in km
    z1d,
    seabed.T,
    cmap=mcolors.ListedColormap([mcolors.to_rgba("tan"), mcolors.to_rgba("white")]),
)

tan_patch = mpatches.Patch(color=mcolors.to_rgba("tan"), label="Land / sea bed")
ax.legend(handles=[tan_patch], loc="lower right")

ax.set_xticks(distances / 1000)

ax.set_ylabel("Depth (m)")
ax.set_xlabel("Distance from start (km)")

plt.colorbar(mesh, ax=ax, label=VARIABLES[plot_variable]["label"])
plt.tight_layout()
```
